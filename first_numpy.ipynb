### Dimensional Arrays in NumPy

- `ndarray` - n-dimensional array

- `0-D array` - single value
- `1-D array` - list (one direction)
- `2-D array` - table (rows & columns)
- `3-D array` - multiple 2D arrays stacked together

```python
import numpy as np
```

```python
# 0-D array
a0 = np.array(5)
print('Value:', a0)
print('Dimension:',a0.ndim)
```

```python
# 1-D array
a1 = np.array([1,2,3])
print('Value:', a1)
print('Dimension:',a1.ndim)
```

```python
#2-D array
a2 = np.array([[1,2,3],
              [4,5,6]])
print('Value:', a2)
print('Dimension:',a2.ndim)
```

```python
#3-D array
a3 = np.array([
    [[1,2],[3,4]],
    [[5,6],[7,8]]])
print('Value:', a3)
print('Dimension:',a3.ndim)
```

```python
# vectorised operations

arr = np.array([1,2,3,4])

print('addition:',arr + 10) 
print('multiplication:',arr * 2) 
print('to the power of:', arr **2)
```

```python
# statistical functions

arr =  np.array([1,4,9,16])
print('Square root:', np.sqrt(arr))
print('Sum:', np.sum(arr))
print('Mean:', np.mean(arr))
print('Standard deviation:', np.std(arr).round(2))
```

```python
# array from 1 to 9

arr = np.arange(1,10)
print('First arr:', arr)

# reshape into 3x3
arr1 = arr.reshape(3,3)
print('Reshaped array:', arr1)

#multiply each element by 5
arr2 = arr * 5
print('Multiply original arr:', arr2)

#find the mean of each row
row_means = np.mean(arr1,axis=1)
print('Mean of rows:', row_means)
```
