# Linear Regression

### Linear Regression Formula

`y = m*x +b`

- `y` - output of value we are predicting
- `x` - input or feature used to make prediction
- `m` - slopw, tells us how much `y` changes when `x` increases
- `b` - intercept, wehre the line crosses the Y-axis when `x=0`

`m = 2` - for every additional hour of studying, the student is expected to score 2 points higher

`b = 1.5` - for students who doesnts study at all, they are expected to score 1.5

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

`X` - features (always 2D array)
- rows = samples
- columns = features

```python
# 2D array - shape(8,1)

# hours of study
X = np.array([[1],[2],[3],[4],[5],[6],[7],[8]])

#score on tests
y = np.array([2,3,5,4,6,5,7,8])
```

```python
df = pd.DataFrame({'Hours_Studied': X.flatten(), 'Grade': y})
plt.scatter(df['Hours_Studied'],df['Grade'],color = 'seagreen')
plt.title('Study Hours vs Grade')
plt.xlabel('Hours Studied')
plt.ylabel('Grade')
plt.grid(True)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X,y)
```

```python
model.predict([[9]])
```

```python
plt.figure(figsize=(8,4))
plt.scatter(X,y,color='purple',label='Actual Data')

x_line = np.linspace(0,9,100).reshape(-1,1)
y_line = model.predict(x_line)

plt.plot(x_line,y_line,color='red',label = 'Model Data')
plt.title('Linear Regression: Study Hours vs Grade')
plt.xlabel('Hours Studied')
plt.ylabel('Grade')
plt.legend()
plt.grid()

```

```python
print('Slope (m):', model.coef_)
print('Intercept (b):', model.intercept_)
```

```python
from sklearn.metrics import mean_squared_error, r2_score
```

```python
y_pred = model.predict(X)
y_pred
```

```python
print('MSE:', mean_squared_error(y,y_pred))
print('R2 Score:', r2_score(y,y_pred))
```

```python
plt.scatter(X,y, color='purple',label = 'Actual Data')
plt.scatter(X,y_pred,color='red',label = 'Predicted Data',marker = 'x')

plt.title('Actual vs Predicted Grade')
plt.xlabel('Hours Studied')
plt.ylabel('Grade')
plt.legend()
plt.grid(True)
```
